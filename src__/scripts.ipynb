# Skewness of target

```python
import numpy as np
from scipy.stats import skew

path_scratch = '/scratch/jczhang/'
path_ds = 'submeso_data_driven/data/mld/'
# U
data = np.load(path_scratch + path_ds + "wT.npz")
wT = data["data"][0,:,:]

y_flat = np.asarray(wT).reshape(-1)
y_flat = y_flat[np.isfinite(y_flat)]

print("Skewness (all pixels, all images):", skew(y_flat, bias=False))
```
